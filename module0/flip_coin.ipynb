```python
import numpy as np
import matplotlib.pyplot as plt

# Số lần tung
n_trials = 1000

# Mô phỏng tung đồng xu: 1 là ngửa, 0 là sấp
results = np.random.choice([0, 1], size=n_trials)

# Tính tổng dồn và tần suất tích lũy
cumulative_sum = np.cumsum(results)
cumulative_freq = cumulative_sum / np.arange(1, n_trials + 1)

# Vẽ
plt.figure(figsize=(10, 6))
plt.plot(np.arange(1, n_trials + 1), cumulative_freq, label='Tần suất mặt ngửa tích lũy')
plt.axhline(y=0.5, color='r', linestyle='--', label='Giá trị lý thuyết (0.5)')
plt.xlabel('Số lần tung')
plt.ylabel('Tần suất')
plt.title('Mô phỏng tung đồng xu 1000 lần')
plt.legend()
plt.grid(True)
plt.show()
```
